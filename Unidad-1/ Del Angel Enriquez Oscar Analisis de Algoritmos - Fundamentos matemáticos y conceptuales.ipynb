**RESPUESTA:*

Un algoritmo es un conjunto finito de instrucciones precisas, reglas operacionales o pasos ordenados lógicamente que permiten resolver un problema de forma sistemática

. Si se visualiza el problema como una función matemática, el algoritmo es el procedimiento mecánico que toma un conjunto de datos de entrada y los transforma en una salida válida

.
Características clave de un algoritmo:
Finitud: Debe terminar siempre después de realizar un número finito de pasos

.
Precisión (No ambigüedad): Los pasos a seguir deben ser claros y unívocos, indicando de forma explícita qué comando o instrucción se debe ejecutar a continuación en cada momento

.
Entrada y Salida: Debe definir con exactitud los datos que requiere para trabajar y los resultados o salidas que va a producir

.
Determinismo: Si se ejecuta el algoritmo múltiples veces con el mismo conjunto de datos de entrada, debe pasar siempre por los mismos estados y producir exactamente el mismo resultado

.
Efectividad: Cada paso o instrucción debe ser lo suficientemente básica como para poder completarse en un tiempo finito

.
Correctitud: Debe cumplir fielmente con el objetivo para el cual fue diseñado, entregando la respuesta correcta para todas las posibles entradas

El Problema de Paro (Halting Problem)

Formulado por el matemático Alan Turing en 1936, el Problema de Paro consiste en determinar si un programa de computadora (o algoritmo) cualquiera, al recibir una entrada determinada, llegará en algún momento a detenerse (finalizar su ejecución) o si, por el contrario, se quedará trabajando indefinidamente (por ejemplo, entrando en un ciclo infinito)

Turing demostró matemáticamente que este es un problema indecidible o insoluble
 Esto significa que es imposible diseñar un algoritmo universal que sea capaz de predecir correctamente si cualquier programa se detendrá o no con una entrada dada

Análisis a Priori vs. Prueba a Posteriori para medir la eficiencia

La gran ventaja del análisis a priori es que es totalmente independiente de factores externos como el hardware, el compilador, el lenguaje de programación o la propia habilidad del programador
. Al ser un cálculo matemático y asintótico, permite a los diseñadores predecir el costo computacional de un algoritmo para entradas muy grandes y descartar opciones ineficientes de manera anticipada, ahorrando un esfuerzo de implementación laborioso e inútil.
Por otro lado, la Prueba a Posteriori (también llamada experimental o empírica) consiste en programar el algoritmo y ejecutarlo en una computadora real utilizando ejemplares o datos de prueba específicos
. Durante esta fase se recogen estadísticas reales y físicas sobre el tiempo de CPU y el espacio en memoria consumidos durante la ejecución.
Su principal fortaleza radica en que proporciona medidas de laboratorio exactas y de alta precisión para los desarrolladores que implementan y usan el software en el mundo real, donde las constantes de proporcionalidad (constantes ocultas) pueden ser críticas para el éxito de una aplicación.

1. Implementa una versión recursiva para el cálculo del factorial ($n! = n \cdot (n-1)!$) y una versión iterativa para Fibonacci.

````python

# FACTORIAL RECURSIVO

def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)

# FIBONACCI ITERATIVO
def fibonacci(n):

    if n == 0:
        return 0
    if n == 1:
        return 1
    a = 0
    b = 1

    for i in range(2, n + 1):
        c = a + b
        a = b
        b = c
    return b

print("Factorial:")
print(factorial(5))
print(factorial(3))

print("\nFibonacci:")
print(fibonacci(5))
print(fibonacci(10))
````

2. Utiliza la librería `time` o la función mágica `%timeit` para comparar el tiempo consumido (A posteriori) entre el Fibonacci recursivo y el iterativo para `n=35`.

````python
import time

# 1. DEFINICION DE LAS FUNCIONES DE FIBONACCI
def fibonacci_recursivo(n):

    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)

def fibonacci_iterativo(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    a = 0
    b = 1
    # Bucle que calcula cada numero una sola vez
    for i in range(2, n + 1):
        c = a + b
        a = b
        b = c

    return b

# 2. MEDICION CON LA LIBRERIA TIME

n = 35
repeticiones_iterativo = 10000
print("COMPARACION DE TIEMPOS PARA FIBONACCI CON n =", n)

# --- 2.1 Medicion del Fibonacci Iterativo ---
print("\n[1] Fibonacci Iterativo (O(n)):")

inicio = time.time()
for _ in range(repeticiones_iterativo):
    resultado_iterativo = fibonacci_iterativo(n)
fin = time.time()

tiempo_total_iterativo = fin - inicio
tiempo_promedio_iterativo = tiempo_total_iterativo / repeticiones_iterativo

print("    Resultado: ", resultado_iterativo)
print("    Tiempo total para", repeticiones_iterativo, "ejecuciones:", tiempo_total_iterativo, "segundos")
print("    Tiempo promedio por ejecucion:", tiempo_promedio_iterativo, "segundos")

# --- 2.2 Medicion del Fibonacci Recursivo ---
print("\n[2] Fibonacci Recursivo (O(phi^n)):")

# El recursivo se ejecuta una sola vez porque es muy lento
inicio = time.time()
resultado_recursivo = fibonacci_recursivo(n)
fin = time.time()

tiempo_recursivo = fin - inicio

print("    Resultado: ", resultado_recursivo)
print("    Tiempo de ejecucion:", tiempo_recursivo, "segundos")

#2.3 Comparacion de los resultados
print("COMPARACION FINAL")
print("Iterativo: ", tiempo_promedio_iterativo, "segundos (promedio)")
print("Recursivo: ", tiempo_recursivo, "segundos")

if tiempo_recursivo > tiempo_promedio_iterativo:
    factor = tiempo_recursivo / tiempo_promedio_iterativo
    print("El iterativo es", factor, "veces mas rapido que el recursivo")
````

### Cuadro Comparativo: Complejidad Temporal Fibonacci Iterativo vs. Recursivo

| Característica | Fibonacci Iterativo (O(n)) | Fibonacci Recursivo (O(φ^n)) |
| :--- | :--- | :--- |
| **Mecanismo** | Bucle (Iteración) | Llamadas a sí mismo (Recursión) |
| **Cálculo de Subproblemas** | Cada número se calcula una sola vez | Mismos subproblemas calculados múltiples veces |
| **Consumo de Recursos** | Bajo y predecible (lineal) | Muy alto, aumenta drásticamente con 'n' (exponencial) |
| **Crecimiento** | Polinomial (O(n)) | Exponencial (O(φ^n), donde φ ≈ 1.618) |
| **Razón de Crecimiento** | Cada paso realiza un número fijo de operaciones y el bucle se ejecuta 'n' veces. | La función se divide en dos llamadas recursivas, creando un árbol de llamadas que duplica el trabajo. |

Como se observa en los resultados de la ejecución anterior (para $n = 35$):
- El **Fibonacci Iterativo** tuvo un tiempo promedio del orden de microsegundos ($O(n)$).
- El **Fibonacci Recursivo** tardó varios segundos ($O(\phi^n)$), demostrando que el enfoque iterativo es cientos de miles de veces más rápido debido a la diferencia entre el crecimiento lineal/polinomial y el exponencial.
